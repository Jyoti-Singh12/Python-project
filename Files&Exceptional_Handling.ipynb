```python
# Q1. Discuss the scenarios where multithreading is preferable to multiprocessing and scenarios where multiprocessing is a better choice

#Answer = 

#Multithreading and multiprocessing are techniques used in Python to run tasks at the same time. The right choice depends on what kind of work the program needs to do.

#Multithreading works best when:

#The program is doing I/O-bound tasks like reading files downloading data or making network calls.

#Several tasks need to share the memory and access the same data.

#Tasks often need to talk to each other during execution.

#We want low memory usage and fast setup for each task.

#The application spends a lot of time waiting for input or output.

#Multiprocessing is better when:

#The program handles CPU-bound tasks, such as math calculations, image editing or processing datasets.

#We have CPU cores and want to use them all.

#Heavy computations need to run at the time.

#Each task can run separately without needing to interact with others.

#We need to get around the limitations of Python’s Global Interpreter Lock (GIL) when working on CPU- jobs.

#Here’s how they differ:

#Multithreading – Uses threads

#Multiprocessing – Uses multiple processes

#Multithreading – Threads share the memory

#Multiprocessing – Each process has its own memory

#Multithreading – Good for I/O-bound tasks

#Multiprocessing – Good for CPU-bound tasks

#Multithreading – Lightweight and quick to start

#Multiprocessing – Uses more system resources

#Multithreading – Can be slowed down by Python’s GIL when doing heavy CPU work

#Multiprocessing – Can use all CPU cores for CPU-heavy tasks

#In short multithreading is usually the better choice for I/O-bound tasks because it’s efficient and uses less memory. Multiprocessing is better, for CPU-bound tasks since it avoids the GIL and uses cores effectively.
```

```python
# Q4. Write a Python program using multithreading where one thread adds numbers to a list, and another
#thread removes numbers from the list. Implement a mechanism to avoid race conditions using
#threading.Lock.

import threading
import time

numbers = []
lock = threading.Lock()


def add_numbers():
    for i in range(1, 11):
        with lock:
            numbers.append(i)
            print("Added:", i)
        time.sleep(0.1)


def remove_numbers():
    for i in range(1, 11):
        while True:
            with lock:
                if numbers:
                    removed = numbers.pop(0)
                    print("Removed:", removed)
                    break
            time.sleep(0.1)



add_thread = threading.Thread(target=add_numbers)
remove_thread = threading.Thread(target=remove_numbers)


add_thread.start()
remove_thread.start()


add_thread.join()
remove_thread.join()

print("Final list:", numbers)
```

```python
# Q6. Discuss why it’s crucial to handle exceptions in concurrent programs and the techniques available for doing so

# Answer :

#Exception handling is very important in programs because multiple threads or processes run at the same time. If an exception occurs in one thread or process and is not handled properly it can cause a task to fail unexpectedly. May affect the reliability of the entire program.

#Why is exception handling important?

#Prevents program crashes: Proper exception handling prevents errors from stopping the entire program.

#Maintains program stability: Other threads or processes can continue working when an error occurs in one task.

#Helps identify errors: Exception handling makes it easier to find and understand problems.

#Protects resources: It helps ensure that files, locks, connections and other resources are properly released.

#Improves reliability: handled errors make concurrent programs more stable and predictable.

#Techniques, for handling exceptions:

#1. Using try-except

#The try-except block is used to catch and handle errors.
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
#2. Using finally

#The finally block is used for cleanup operations and executes whether an exception occurs or not.
try:
    print("Executing task")
except Exception as e:
    print("Error:", e)
finally:
    print("Task completed")
```

```python
#3. Handling exceptions with Futures

#When using ThreadPoolExecutor or ProcessPoolExecutor, exceptions can be caught when calling future.result().

from concurrent.futures import ThreadPoolExecutor

def divide(a, b):
    return a / b

with ThreadPoolExecutor(max_workers=2) as executor:
    future = executor.submit(divide, 10, 0)

    try:
        print(future.result())
    except Exception as e:
        print("Error:", e)
```

```python
# Q7. Create a program that uses a thread pool to calculate the factorial of numbers from 1 to 10 concurrently.
#Use concurrent.futures.ThreadPoolExecutor to manage the threads.
from concurrent.futures import ThreadPoolExecutor
import math


def calculate_factorial(n):
    return n, math.factorial(n)


numbers = range(1, 11)

with ThreadPoolExecutor(max_workers=5) as executor:
    results = executor.map(calculate_factorial, numbers)

for number, factorial in results:
    print(f"{number}! = {factorial}")
```

```python
# Q8. Create a Python program that uses multiprocessing.Pool to compute the square of numbers from 1 to 10 in
#parallel. Measure the time taken to perform this computation using a pool of different sizes (e.g., 2, 4, 8 processes).
import os
import sys
import tempfile

worker_code = r'''
import multiprocessing
import time
import sys

def square(n):
    return n * n

if __name__ == "__main__":
    processes = int(sys.argv[1])
    numbers = range(1, 11)

    start_time = time.perf_counter()

    with multiprocessing.Pool(processes=processes) as pool:
        results = pool.map(square, numbers)

    end_time = time.perf_counter()

    print(f"Processes: {processes}")
    print(f"Squares: {results}")
    print(f"Time taken: {end_time - start_time:.6f} seconds")
'''

worker_file = os.path.join(
    tempfile.gettempdir(),
    "multiprocessing_square.py"
)

with open(worker_file, "w") as file:
    file.write(worker_code)

print("Worker program created successfully.")
```

```python
import subprocess

for processes in [2, 4, 8]:
    result = subprocess.run(
        [sys.executable, worker_file, str(processes)],
        capture_output=True,
        text=True
    )

    print(result.stdout)

    if result.stderr:
        print("Error:", result.stderr)
```
