```python
#Q1. What is Abstraction in OOPs? Explain with an example

#Abstraction is an OOP concept of hiding unnecessary implementation details and showing only the essential features to the user.

#For example, when we use an ATM, we simply enter our PIN and withdraw money. We do not need to know the internal programming or banking operations performed by the ATM.

from abc import ABC, abstractmethod 
class Animal(ABC): 
    @abstractmethod 
    def sound(self): 
        pass 
class Dog(Animal): 
    def sound(self): 
        print("Dog barks") 
dog = Dog() 
dog.sound()
```

```python
#Q2. Differentiate between Abstraction and Encapsulation. Explain with an example.

#Both abstraction and encapsulation are important concepts of OOP, but they have different purposes.

   #Abstraction	
#Hides implementation details.	
#Focuses on what an object does.	
#Reduces complexity for the user.	
#Commonly implemented using abstract classes and methods.	
#Example


class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

account = BankAccount(5000)

print(account.get_balance())




```

```python
  #Encapsulation
#Bundles data and methods together.

#Focuses on controlling how data is accessed.

#Protects data from inappropriate access.

#Commonly implemented using classes, private attributes, and properties.

#Example: Keeping an account balance private.

```

```python
#Q3. What is abc module in Python? Why is it used?
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Dog barks")

d = Dog()
d.sound()
```

```python
#Q4. How can we achieve Data Abstraction?

#In Python, data abstraction can be achieved using abstract classes and abstract methods provided by the abc module.

from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):

    def area(self):
        radius = 5
        return 3.14 * radius * radius

c = Circle()
print(c.area())
```

```python
#Q5. Can we create an instance of an abstract class? Explain your answer.

#No, we cannot directly create an instance of an abstract class if it contains one or more abstract methods.

# example:

from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

  
class Dog(Animal):

    def sound(self):
        print("Dog barks")

d = Dog()
d.sound()
```
