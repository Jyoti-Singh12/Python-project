```python
#Q1. Creating a DataFrame
import pandas as pd

data = {
    "Product": ["Laptop", "Tablet", "Phone", "Monitor", "Keyboard"],
    "Region": ["West", "South", "East", "North", "West"],
    "Sales": [1056, 904, 2028, 1269, 777],
    "Quantity": [2, 7, 4, 8, 8]
}

df_q1 = pd.DataFrame(data)

print(df_q1)
```

```python
#Q2. Data Inspection
df = pd.read_csv("Retail_Dataset.csv")

print("First 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nDataset Information:")
df.info()

print("\nShape of Dataset:")
print(df.shape)
```

```python
#Q3. Column Selection
selected_columns = df[["Product", "Region", "Sales"]]

display(selected_columns)
```

```python
#Q4. Indexing using .loc
print("Row with index 10:")
display(df.loc[10])

print("\nSales value of row with index 20:")
print(df.loc[20, "Sales"])
```

```python
#Q5. Indexing using .iloc
print("First 6 rows and first 4 columns:")
display(df.iloc[:6, :4])

print("\nSales value from the 3rd row:")
print(df.iloc[2]["Sales"])
```

```python
#Q6. Row Slicing
print("First 10 rows:")
display(df.iloc[:10])

print("\nRows 15 to 25:")
display(df.iloc[15:26])
```

```python
#Q7. Filtering Data
print("Rows where Region = East:")
east_data = df[df["Region"] == "East"]
display(east_data)

print("\nRows where Sales > 1500:")
high_sales = df[df["Sales"] > 1500]
display(high_sales)
```

```python
#Q8. Multiple Condition Filtering
filtered_data = df[
    (df["Region"] == "West") &
    (df["Sales"] > 1200)
]

display(filtered_data)
```

```python
#Q9. Descriptive Statistics
print("Summary Statistics:")
display(df.describe())
```

```python
mean_sales = df["Sales"].mean()
max_quantity = df["Quantity"].max()
min_sales = df["Sales"].min()

print("Mean Sales:", mean_sales)
print("Maximum Quantity Sold:", max_quantity)
print("Minimum Sales:", min_sales)
```

```python
#Q10. Aggregation Analysis
total_sales = df["Sales"].sum()
average_sales = df["Sales"].mean()
maximum_sales = df["Sales"].max()
minimum_sales = df["Sales"].min()

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Maximum Sales:", maximum_sales)
print("Minimum Sales:", minimum_sales)
```
