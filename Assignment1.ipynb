```python
#Write a code to count the number of vowels in a string

def count_vowels(s):
    count = 0

    for char in s.lower():
        if char in "aeiou":
            count += 1

    return count


text = "Hello World"
result = count_vowels(text)

print("Number of vowels:", result)
```

```python
#Write a code to check if a given string is a palindrome or not
def check_palindrome(s):
    if s == s[::-1]:
        return "Palindrome"
    else:
        return "Not a Palindrome"


text = "madam"
result = check_palindrome(text)

print(result)
```

```python
def check_anagram(str1, str2):
    if sorted(str1.lower()) == sorted(str2.lower()):
        return "Anagrams"
    else:
        return "Not Anagrams"


string1 = "listen"
string2 = "silent"

result = check_anagram(string1, string2)
print(result)
```

```python
#Write a code to find all occurrences of a given substring within another string
def find_occurrences(string, substring):
    positions = []

    for i in range(len(string) - len(substring) + 1):
        if string[i:i + len(substring)] == substring:
            positions.append(i)

    return positions


text = "hello hello hello"
substring = "hello"

result = find_occurrences(text, substring)

print("Occurrences found at:", result)
```

```python
#Write a code to perform basic string compression using the counts of repeated characters
def compress_string(s):
    result = ""
    count = 1

    for i in range(len(s) - 1):
        if s[i] == s[i + 1]:
            count += 1
        else:
            result += s[i] + str(count)
            count = 1

    result += s[-1] + str(count)

    return result


text = "aaabbcccc"
result = compress_string(text)

print(result)
```

```python
#Write a code to reverse a list in-place without using any built-in reverse functions
def reverse_list(my_list):
    left = 0
    right = len(my_list) - 1

    while left < right:
        my_list[left], my_list[right] = my_list[right], my_list[left]

        left += 1
        right -= 1


numbers = [1, 2, 3, 4, 5]

reverse_list(numbers)

print(numbers)
```

```python
#Implement a code to find and remove duplicates from a list while preserving the original order of elements
def remove_duplicates(my_list):
    result = []

    for element in my_list:
        if element not in result:
            result.append(element)

    return result


numbers = [1, 2, 2, 3, 1, 4, 3, 5]

result = remove_duplicates(numbers)

print(result)
```

```python
#Create a code to check if a given list is sorted (either in ascending or descending order) or not
def check_sorted(my_list):
    ascending = my_list == sorted(my_list)
    descending = my_list == sorted(my_list, reverse=True)

    if ascending:
        return "List is sorted in ascending order"
    elif descending:
        return "List is sorted in descending order"
    else:
        return "List is not sorted"


numbers = [1, 2, 3, 4, 5]

result = check_sorted(numbers)

print(result)
```

```python
#Write a code to merge two sorted lists into a single sorted list
def merge_sorted_lists(list1, list2):
    result = []
    i = 0
    j = 0

    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            result.append(list1[i])
            i += 1
        else:
            result.append(list2[j])
            j += 1

    while i < len(list1):
        result.append(list1[i])
        i += 1

    while j < len(list2):
        result.append(list2[j])
        j += 1

    return result


list1 = [1, 3, 5, 7]
list2 = [2, 4, 6, 8]

result = merge_sorted_lists(list1, list2)

print(result)
```

```python
#Implement a code to find the intersection of two given lists
def find_intersection(list1, list2):
    result = []

    for element in list1:
        if element in list2 and element not in result:
            result.append(element)

    return result


list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

result = find_intersection(list1, list2)

print(result)
```

```python
#Create a code to find the union of two lists without duplicates

def find_union(list1, list2):
    result = []

    for element in list1:
        if element not in result:
            result.append(element)

    for element in list2:
        if element not in result:
            result.append(element)

    return result


list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]

result = find_union(list1, list2)

print(result)
```

```python
#Write a code to shuffle a given list randomly without using any built-in shuffle functions
import random

def shuffle_list(my_list):
    for i in range(len(my_list) - 1, 0, -1):
        j = random.randint(0, i)
        my_list[i], my_list[j] = my_list[j], my_list[i]

    return my_list


numbers = [1, 2, 3, 4, 5]

result = shuffle_list(numbers)

print(result)
```

```python
#Write a code that takes two tuples as input and returns a new tuple containing elements that are common to both input tuples
def common_elements(tuple1, tuple2):
    result = []

    for element in tuple1:
        if element in tuple2 and element not in result:
            result.append(element)

    return tuple(result)


tuple1 = (1, 2, 3, 4, 5)
tuple2 = (3, 4, 5, 6, 7)

result = common_elements(tuple1, tuple2)

print(result)
```

```python
#Write a code that takes a list of words as input and returns a dictionary where the keys are unique words and the values are the frequencies of those words in the input list
def word_frequency(words):
    frequency = {}

    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    return frequency


words = ["apple", "banana", "apple", "orange", "banana", "apple"]

result = word_frequency(words)

print(result)
```

```python
#Write a code that takes two dictionaries as input and merges them into a single dictionary. If there are common keys, the values should be added together
def merge_dictionaries(dict1, dict2):
    result = dict1.copy()

    for key, value in dict2.items():
        if key in result:
            result[key] += value
        else:
            result[key] = value

    return result


dict1 = {"a": 10, "b": 20, "c": 30}
dict2 = {"b": 5, "c": 15, "d": 40}

result = merge_dictionaries(dict1, dict2)

print(result)
```

```python
#Write a code to access a value in a nested dictionary. The function should take the dictionary and a list of keys as input, and return the corresponding value. If any of the keys do not exist in the dictionary, the function should return None
def get_nested_value(dictionary, keys):
    current = dictionary

    for key in keys:
        if key not in current:
            return None

        current = current[key]

    return current


data = {
    "student": {
        "details": {
            "name": "Jyoti",
            "age": 21
        }
    }
}

keys = ["student", "details", "name"]

result = get_nested_value(data, keys)

print(result)
```

```python
#Write a code that takes a dictionary as input and returns a sorted version of it based on the values. You can choose whether to sort in ascending or descending order
def sort_dictionary(dictionary):
    return dict(sorted(dictionary.items(), key=lambda item: item[1]))


data = {
    "apple": 50,
    "banana": 20,
    "orange": 40,
    "mango": 30
}

result = sort_dictionary(data)

print(result)
```

```python
#Write a code that inverts a dictionary, swapping keys and values. Ensure that the inverted dictionary correctly handles cases where multiple keys have the same value by storing the keys as a list in the inverted dictionary
def invert_dictionary(dictionary):
    inverted = {}

    for key, value in dictionary.items():
        if value in inverted:
            inverted[value].append(key)
        else:
            inverted[value] = [key]

    return inverted


data = {
    "a": 1,
    "b": 2,
    "c": 1,
    "d": 3,
    "e": 2
}

result = invert_dictionary(data)

print(result)
```
