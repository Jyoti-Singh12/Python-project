```python
#Q1. Load dataset and display first 5 rows + information
import pandas as pd
import numpy as np

df = pd.read_csv("retail_sales.csv")

print("First 5 rows:")
print(df.head())

print("\nDataset Information:")
df.info()
```

```python
#Q2. Identify missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
#Q3. Replace missing Quantity and Price with their mean
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Price"] = df["Price"].fillna(df["Price"].mean())

print(df[["Quantity", "Price"]].head())
```

```python
#Q4. Remove rows with missing Product Category or Region
df = df.dropna(subset=["Product Category", "Region"])

print("Dataset after removing missing Product Category or Region:")
print(df)
```

```python
#Q5. Create Revenue column using NumPy

df["Revenue"] = np.multiply(df["Quantity"], df["Price"])

print(df[["Transaction ID", "Product Category", "Quantity", "Price", "Revenue"]])
```

```python
#Q6. Calculate total revenue
total_revenue = np.sum(df["Revenue"])

print("Total Revenue:", total_revenue)
```

```python
#Q7. Total revenue by Product Category
category_revenue = df.groupby("Product Category")["Revenue"].sum()

print("Revenue by Product Category:")
print(category_revenue)
```

```python
#Q8. Top 3 and Bottom 3 categories
top_3 = category_revenue.head(3)
bottom_3 = category_revenue.tail(3)

print("Top 3 Categories:")
print(top_3)

print("\nBottom 3 Categories:")
print(bottom_3)
```

```python
#Q9. Revenue by Region
region_revenue = df.groupby("Region")["Revenue"].sum()

print("Revenue by Region:")
print(region_revenue)
```

```python
# Q10. Monthly revenue + Mean, Median, Standard Deviation


df["Month"] = df["Date"].dt.month_name()

monthly_revenue = df.groupby("Month")["Revenue"].sum()

print(monthly_revenue)
```

```python
# Mean 
mean_revenue = np.mean(df["Revenue"])
print("Mean Revenue:", mean_revenue)
```

```python
# Median
median_revenue = np.median(df["Revenue"])
print("Median Revenue:", median_revenue)
```

```python
#Standard Deviation
std_revenue = np.std(df["Revenue"])

print("Standard Deviation:", std_revenue)
```
