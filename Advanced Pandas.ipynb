```python
import pandas as pd

#Q1. Dataset Exploration

# Load dataset
df = pd.read_csv("Advanced_pandas_sales_dataset.csv")

# First 10 rows
print("First 10 rows:")
display(df.head(10))

# Dataset structure
print("\nDataset Information:")
df.info()

# Number of rows and columns
print("\nNumber of rows and columns:")
print(df.shape)

# Data types
print("\nData Types:")
print(df.dtypes)
```

```python
#Q2. Sales Filtering Analysis
# Orders where Sales > 3500
sales_above_3500 = df[df["Sales"] > 3500]

print("Orders where Sales > 3500:")
display(sales_above_3500)

# East region and Sales > 2500
east_high_sales = df[
    (df["Region"] == "East") &
    (df["Sales"] > 2500)
]

print("\nEast Region with Sales > 2500:")
display(east_high_sales)
```

```python
#Q3. Customer Purchase Analysis
customer_sales = (
    df.groupby("Customer")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 5 Customers based on Total Sales:")
display(customer_sales.head(5))
```

```python
#Q4. Regional Sales Performance

region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Total Sales by Region:")
display(region_sales)

highest_region = region_sales.idxmax()
highest_region_sales = region_sales.max()

print("\nRegion with Highest Revenue:", highest_region)
print("Highest Revenue:", highest_region_sales)
```

```python
#Q5. Product Performance Analysis

product_quantity = (
    df.groupby("Product")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("Total Quantity Sold by Product:")
display(product_quantity)

most_popular_product = product_quantity.idxmax()
highest_quantity = product_quantity.max()

print("\nMost Popular Product:", most_popular_product)
print("Total Quantity Sold:", highest_quantity)
```

```python
#Q6. Multi-Level Grouping

region_category_sales = (
    df.groupby(["Region", "Category"])["Sales"]
    .sum()
)

print("Total Sales by Region and Category:")
display(region_category_sales)
```

```python
#Q7. Pivot Table Creation

pivot_table = pd.pivot_table(
    df,
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="sum",
    fill_value=0
)

print("Regional Product Sales Pivot Table:")
display(pivot_table)
```

```python
#Q8. Profitability Analysis

category_profit = (
    df.groupby("Category")["Profit"]
    .mean()
    .sort_values(ascending=False)
)

print("Average Profit by Category:")
display(category_profit)
```

```python
#Q9. Discount Impact Study
discount_profit = (
    df.groupby("Discount")["Profit"]
    .mean()
    .sort_index()
)

print("Average Profit by Discount Level:")
display(discount_profit)
```

```python
#Q10. End-to-End Data Wrangling
# Step 1: Filter Sales > 3000
filtered_df = df[df["Sales"] > 3000]

print("Orders where Sales > 3000:")
display(filtered_df)
```

```python
#Step 2: Group by Region and Product
grouped_sales = (
    filtered_df
    .groupby(["Region", "Product"])["Sales"]
    .sum()
    .reset_index()
)

print("Total Sales by Region and Product:")
display(grouped_sales)
```

```python
#Step 3: Create Pivot Table
final_pivot = grouped_sales.pivot(
    index="Region",
    columns="Product",
    values="Sales"
).fillna(0)

print("Final Pivot Table:")
display(final_pivot)
```
