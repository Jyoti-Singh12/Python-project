```python
#Q1. Explain Class and Object with respect to Object-Oriented Programming. Give a suitable example.

#Class

#A class is a blueprint or template used to create objects. It defines the attributes (data) and methods (functions) that an object will have.

#For example, a Student class can contain a student's name, age, and a method to display the student's information.

#Object

#An object is an instance of a class. It represents a real-world entity and has its own data.


class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)


student1 = Student("Jyoti", 21)
student1.display()
```

```python
#Q2. Name the four pillars of OOPs.

#The four main pillars of Object-Oriented Programming are:

#Encapsulation – Bundling data and methods together and controlling access to data.

#Inheritance – Creating a new class from an existing class.

#Polymorphism – Allowing the same method or interface to behave differently for different objects.

#Abstraction – Hiding unnecessary implementation details and showing only essential information.
```

```python
#Q3. Explain why the __init__() function is used. Give a suitable example.

#The __init__() function is a special method in Python classes. It is automatically called when an object is created.

#It is mainly used to initialize the attributes of an object.

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Jyoti", 21)

print("Name:", student1.name)
print("Age:", student1.age)
```

```python
#Q4. Why is self used in OOPs?

#self is used in Python classes to refer to the current object.

class Student: 
    def __init__(self, name, age): 
        self.name = name
        self.age = age 
    def display(self): 
        print("Name:", self.name) 
        print("Age:", self.age) 
student1 = Student("Sakshi", 21) 
student1.display()
```

```python
#Q5. What is inheritance? Give an example for each type of inheritance.

#Inheritance

#Inheritance is an OOP concept in which a new class, called the child class (derived class), acquires the properties and methods of an existing class, called the parent class (base class)

#1. Single Inheritance

#In single inheritance, one child class inherits from one parent class.

#2. Multiple Inheritance

#In multiple inheritance, one child class inherits from more than one parent class.

#3. Multilevel Inheritance

#In multilevel inheritance, a class inherits from another derived class, forming a chain.

#4. Hierarchical Inheritance

#In hierarchical inheritance, multiple child classes inherit from the same parent class.

#5. Hybrid Inheritance

#Hybrid inheritance is a combination of two or more types of inheritance.

```
