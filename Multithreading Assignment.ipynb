```python
#Q1. What is Multithreading in Python? Why is it used? Name the module used to handle threads in Python.

#Multithreading is a technique in which multiple threads run concurrently within a single program.

#A thread is a small unit of execution. Using multiple threads allows a program to perform multiple tasks concurrently, especially when tasks spend time waiting for I/O operations such as file operations, network requests, or user input.

#Why is multithreading used?

#Multithreading is used to:

#Perform multiple tasks concurrently.

#Improve responsiveness of applications.

#Efficiently handle I/O-bound tasks.

#Perform background tasks while the main program continues running.

#Reduce the waiting time between tasks.

#Module used

import threading

def task():
    print("Thread is running")

t = threading.Thread(target=task)
t.start()
```

```python
#Q3. Explain the following functions: run(), start(), join(), isAlive()

#1. run()

import threading

def display():
    print("Hello from thread")

t = threading.Thread(target=display)
t.start()
```

```python
#2. start()

import threading

def task():
    print("Task is running")

t = threading.Thread(target=task)
t.start()
```

```python
#3. join()

import threading
import time

def task():
    time.sleep(2)
    print("Task completed")

t = threading.Thread(target=task)

t.start()
t.join()

print("Main thread continues")
```

```python
#4. isAlive()

import threading

def task():
    print("Task is running")

t = threading.Thread(target=task)

print(t.is_alive())

t.start()

print(t.is_alive())
```

```python
#Q4. Write a Python program to create two threads. Thread one must print the list of squares and thread two must print the list of cubes

import threading

numbers = [1, 2, 3, 4, 5]

def print_squares():
    squares = [x ** 2 for x in numbers]
    print("Squares:", squares)

def print_cubes():
    cubes = [x ** 3 for x in numbers]
    print("Cubes:", cubes)

thread1 = threading.Thread(target=print_squares)
thread2 = threading.Thread(target=print_cubes)

thread1.start()
thread2.start()

thread1.join()
thread2.join()

print("Both threads completed.")
```

```python
#Q5. State the advantages and disadvantages of Multithreading

#Advantages

#Concurrent execution – Multiple tasks can make progress at the same time.

#Better responsiveness – Applications can remain responsive while background tasks run.

#Efficient for I/O-bound tasks – Useful for tasks involving files, networks, databases, etc.

#Resource sharing – Threads within the same process can share memory and resources.

#Lower overhead – Threads generally require fewer resources than separate processes.

#Disadvantages

#Race conditions – Multiple threads accessing shared data can cause incorrect results.

#Deadlocks – Threads can become stuck waiting for each other.

#ifficult debugging – Thread-related bugs can be difficult to reproduce.

#ynchronization overhead – Locks and other synchronization mechanisms add complexity.

#Limited CPU parallelism in standard CPython – The Global Interpreter Lock (GIL) generally prevents multiple Python threads from executing Python bytecode simultaneously in the same interpreter, so threads are not usually the way to speed up CPU-bound Python code.
```
