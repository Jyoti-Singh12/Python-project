```python
#Q2. What happens when an exception is not handled?

#When an exception is not handled, Python stops the execution of the program and displays an error message (traceback).


num = 10
den = 0

result = num / den

print("Program continues")
```

```python
#Q3. Which Python statements are used to catch and handle exceptions?

try:
    num = int(input("Enter a number: "))
    result = 10 / num

except ZeroDivisionError:
    print("Cannot divide by zero")

except ValueError:
    print("Please enter a valid number")
```

```python
#Q4. Explain with an example: try, else, finally, raise

#Try

try:
    x = 10 / 2
    print(x)
except:
    print("An error occurred")
```

```python
#2. else
try:
    x = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Division successful")
```

```python
#3. finally
try:
    x = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("This block always executes")
    
```

```python
#4. raise

age = 15

if age < 18:
    raise ValueError("Age must be 18 or above")
    
```

```python
#Q5. What are Custom Exceptions in Python? Why do we need Custom Exceptions?

#Custom Exceptions are user-defined exceptions created by the programmer for handling specific situations or application-specific errors.

# Why do we need Custom Exceptions?
 
# Custom exceptions are useful because they:

# Make error handling more meaningful.

# Allow us to define application-specific errors.

class InsufficientBalanceError(Exception):
    pass


balance = 1000
withdraw = 1500

if withdraw > balance:
    raise InsufficientBalanceError("Insufficient balance")
```

```python
#Q6. Create a custom exception class. Use this class to handle an exception

class AgeError(Exception):
    pass


try:
    age = int(input("Enter your age: "))

    if age < 18:
        raise AgeError("Age must be 18 or above")

    print("You are eligible.")

except AgeError as e:
    print("Custom Exception:", e)
```
